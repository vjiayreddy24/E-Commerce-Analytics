```python
files = dbutils.fs.ls("/Volumes/databrickspractice/bronze/bronze_volume/brazil-ecom-data/")
for f in files:
    if f.name.startswith("olist"):
        dbutils.fs.mv(f.path, "/Volumes/databrickspractice/bronze/bronze_volume/brazil-ecom-data/" + f.name)
```

```python
files_to_move = ["/Volumes/databrickspractice/bronze/bronze_volume/product_category_name_translation.csv"]
for file_path in files_to_move:
    file_name = [file_path.split("/")[-1]]
    try:
        dbutils.fs.ls(file_path) # will fail when there no file exits
        dbutils.fs.mv(file_path, f"/Volumes/databrickspractice/bronze/bronze_volume/{file_name}")
        print(f"File {file_path} moved successfully.")
    except:
        print(f"File {file_path} does not exist.")

```

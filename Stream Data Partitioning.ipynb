```python
incremental_orders = spark.read.format("csv").option("header", "true").load("/Volumes/databrickspractice/bronze/bronze_volume/brazil-ecom-data/olist_orders_dataset.csv")
display(increamental_orders)
```

```python
from pyspark.sql.functions import *
from delta.tables import DeltaTable

target_table = "Volumes/databrickspractice/bronze/bronze_volume/brazil-ecom-data/stream_data/"

incremental_orders= incremental_orders.withColumn("order_year", year(col("order_purchase_timestamp"))) \
                                       .withColumn("order_month", month(col("order_purchase_timestamp")))
```

```python
incremental_orders.write.format("csv") \
            .mode("overwrite").option("header", "true") \
            .partitionBy("order_year", "order_month") \
            .save(target_table)
```
